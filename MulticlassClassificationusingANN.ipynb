```python
df = pd.read_csv("df.csv")
le=LabelEncoder()
x=df.drop("Class",axis=1)
y=df["Class"]

y=le.fit_transform(y)
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
se=StandardScaler()
x_train_scaled=se.fit_transform(x_train)
x_test_scaled=se.transform(x_test)
```

## **ANN**

```python
# Tensors
X_train_tensors=torch.tensor(x_train_scaled,dtype=torch.float32)
y_train_tensors=torch.tensor(y_train,dtype=torch.long)

X_test_tensors=torch.tensor(x_test_scaled,dtype=torch.float32)
y_test_tensors=torch.tensor(y_test,dtype=torch.long)


```

```python
# dataset
train_dataset=torch.utils.data.TensorDataset(X_train_tensors,y_train_tensors)
test_dataset=torch.utils.data.TensorDataset(X_test_tensors,y_test_tensors)
```

```python
# dataloader class
train_dataloader=torch.utils.data.DataLoader(train_dataset,batch_size=,32shuffle=True)
test_dataloader=torch.utils.data.DataLoader(test_dataset,batch_size=32,shuffle=False)
```

```python
# model
class Ann(nn.Module):
  def __init__(self):
    super(Ann,self).__init__()
    self.model=nn.Sequential(
        nn.Linear(x.shape[1],64),
        nn.ReLU(),
        nn.Linear(64,64),
        nn.ReLU(),
        nn.Linear(64,7),
    )
  def forward(self,x):
    return self.model(x)
```

```python
model=Ann()
criterion=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters())
```

```python
epochs=100
for epoch in range(epochs):
  model.train()
  running_loss=0.0
  for xb,yb in train_dataloader:
    output=model(xb)
    loss=criterion(output,yb)
    loss.backward()
    optimizer.step()
    running_loss+=loss.item()

  epoch_loss=running_loss/len(train_dataloader)

  print(f"for {epoch/epochs} ==>>  loss:{running_loss} ")
```

```python
# evaluation
total=0
correct=0
model.eval()
with torch.no_grad():
  for xb,yb in test_dataloader:
    output=model(xb)
    _,predicted=torch.max(output,1)
    correct+=(predicted==yb).sum().item()
    total+=yb.size(0)



  print(total)
  print(correct)
```

```python
accuracy=correct/total*100
print(accuracy)
```
